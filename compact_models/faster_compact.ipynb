```python
import torch
import torch.nn as nn
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor

def modify_faster_rcnn_layers(model, num_classes=3):
    """
    Modifica apenas as camadas finais de um modelo existente
    """
    # Obtém o número de features de entrada
    in_features = model.roi_heads.box_predictor.cls_score.in_features
    
    # Modifica a camada de classificação
    # +1 para incluir a classe background (índice 0)
    model.roi_heads.box_predictor.cls_score = nn.Linear(in_features, num_classes + 1)
    
    # Modifica a camada de regressão de bounding box
    # 4 coordenadas (x, y, w, h) para cada classe (exceto background)
    model.roi_heads.box_predictor.bbox_pred = nn.Linear(in_features, num_classes * 4)
    
    return model

model = modify_faster_rcnn_layers(model, num_classes=3)

# Verificar a arquitetura modificada
print("Classificador modificado:")
print(f"cls_score: {model.roi_heads.box_predictor.cls_score}")
print(f"bbox_pred: {model.roi_heads.box_predictor.bbox_pred}")
```
